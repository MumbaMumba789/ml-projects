# Iris Species Classifier

Predicts the species of an iris flower (setosa, versicolor, virginica) from
its sepal and petal measurements, using a Decision Tree classifier.

```python
import joblib
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

RANDOM_STATE = 42
```

## Load and inspect the data

```python
df = pd.read_csv('Iris.csv')
print(df.head())
print(df.shape)
print("\nMissing values:\n", df.isnull().sum())
```

## Split features/target and train/test sets

```python
FEATURE_COLUMNS = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
TARGET_COLUMN = 'Species'

X = df[FEATURE_COLUMNS]
y = df[TARGET_COLUMN]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y,  # keep species proportions balanced across the split
)
```

## Train the model

```python
model = DecisionTreeClassifier(random_state=RANDOM_STATE)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
y_pred
```

## Evaluate

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

## Save the trained model

```python
joblib.dump(model, 'iris_model.joblib')
print("Model saved to iris_model.joblib")
```
